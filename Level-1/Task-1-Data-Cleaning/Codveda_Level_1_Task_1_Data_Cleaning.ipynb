## Data Cleaning and preprocessing

### Project Objective

The objective of this project is to clean and preprocess the provided sentiment dataset by identifying and handling missing values, removing duplicates, standardizing inconsistent data formats, and ensuring that variables have appropriate data types for further analysis.

### Dataset Overview
The original dataset contains 732 rows and 15 columns. The dataset contains information about social media posts, including sentiment, timestamp, platform, hashtags, engagement metrics and country.

### Cleaning Process
The following preprocessing steps were performed:

1. Inspected the structure and data types of the dataset.
2. Checked for missing values.
3. Identified and removed unnecessary index columns.
4. Identified and removed duplicate records.
5. Removed leading and trailing whitespace from text-based columns.
6. Converted the `Timestamp` column to datetime format.
7. Validated the cleaned dataset.


### Final Result
After preprocessing, the dataset contained 710 rows and 13 columns, with no missing values or duplicate records.

## 5. Removing unnecessary columns

The dataset contained two columns `Unnamed: 0` and `Unnamed: 0.1`, which were identified as redundant index-tracking columns rather than meaningful analytical variables.

The `Timestamp` column was initially stored as an object(text) data type. Since it represents date and time information, it was converted to pandas' datetime format.

## Text Standardization

Leading and trailing whitespace was removed from the relevant text-based columns to standardize categorical and textual values.

After cleaning, the number of unique values in the `Sentiment` column decreased from 279 to 191, indicating that whitespace inconsistencies had caused multple representations of the otherwise identical values. 

```python
text_columns = ["Text", "Sentiment", "User", "Platform", "Hashtags", "Country"]

for column in text_columns:
 df[column] = df[column].str.strip()
```
